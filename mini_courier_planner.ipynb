# Mini-Courier Planner
This notebook demonstrates the integrated capstone pipeline: merge sort, 0/1 knapsack DP, Dijkstra shortest paths, and nearest-neighbor TSP approximation.

```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd()))
from src.io_utils import load_parcels, load_graph
from src.algorithms import merge_sort_parcels, knapsack, dijkstra, nearest_neighbor_route
parcels = load_parcels('datasets/parcels.csv')
graph = load_graph('datasets/routes.json')
```

## Phase 2: prioritization and box filling

```python
prioritized = merge_sort_parcels(parcels)
[(p.parcel_id, p.priority, p.deadline) for p in prioritized]
```

```python
selected, total_value, dp_table = knapsack(prioritized, capacity=15)
print('Selected:', [p.parcel_id for p in selected])
print('Weight:', sum(p.weight for p in selected), 'Value:', total_value)
print('Final DP row:', dp_table[-1])
```

## Phase 3: shortest paths and route approximation

```python
distances, previous = dijkstra(graph, 'Depot')
print('Shortest distances from Depot:')
print(distances)
```

```python
stops = [p.destination for p in selected]
route = nearest_neighbor_route(graph, 'Depot', stops)
print('Stops:', stops)
print('Route:', ' -> '.join(route.path))
print('Distance:', route.distance)
```
